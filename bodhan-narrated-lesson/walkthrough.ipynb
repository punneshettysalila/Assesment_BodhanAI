# Bodhan narrated lesson walkthrough

This notebook explains the scanned-chapter pipeline and runs only the offline, API-free demo.

## Pipeline

The pipeline ingests a PDF or images, performs OCR and layout, translates to target languages, narrates sections, verifies speech, and writes HTML/JSON reports. Offline mode replaces all models with deterministic simulations.

```python
from pathlib import Path
import sys, json, subprocess

ROOT = Path.cwd()
OUT_DIR = globals().get("OUT_DIR")
out = Path(OUT_DIR) if OUT_DIR else ROOT / "out" / "notebook_demo"
if not (out / "report.json").exists():
    subprocess.run([sys.executable, str(ROOT / "tools" / "offline_demo.py"), "--out", str(out)], check=True)
print(out)
```

## report.json

```python
report = json.loads((out / "report.json").read_text(encoding="utf-8"))
rows = []
for lang, data in report["languages"].items():
    rows.append({"Language": lang, "Status": data["status"], "Structure": data.get("structure_ok")})
import pandas as pd
pd.DataFrame(rows)
```
